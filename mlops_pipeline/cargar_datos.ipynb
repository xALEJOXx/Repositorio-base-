# Carga de la base de créditos

Este notebook carga el archivo Excel original y genera la base en
formato CSV requerida para el proyecto. Se verifican sus dimensiones
y columnas, conservando los valores originales. La limpieza y el
análisis exploratorio se realizarán en comprension_eda.ipynb.

```python
from pathlib import Path
import pandas as pd

ROOT = Path.cwd()
if ROOT.name == "mlops_pipeline":
    ROOT = ROOT.parent

archivo_excel = ROOT / "BD_creditos (2).xlsx"
archivo_csv = ROOT / "Base_de_datos.csv"

datos = pd.read_excel(archivo_excel, sheet_name="Hoja1")

print("Filas:", datos.shape[0])
print("Columnas:", datos.shape[1])
display(datos.head())

datos.to_csv(archivo_csv, index=False, encoding="utf-8-sig")

verificacion = pd.read_csv(archivo_csv)
assert verificacion.shape == datos.shape
assert verificacion.columns.tolist() == datos.columns.tolist()

print("CSV generado y verificado correctamente.")
```
